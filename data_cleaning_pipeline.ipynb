```python
# Import the pandas library
import pandas as pd

# Load the messy dataset into a DataFrame
# Make sure the file name matches exactly what you downloaded
df = pd.read_csv('Snitch_Fashion_Sales_Uncleaned.csv')

# Display the first 5 rows to see what the data looks like
print("First 5 rows of the data:")
display(df.head())

# Get a summary of the data, including data types and missing values
print("\nData summary and info:")
df.info()
```

```python
# Convert 'Order_Date' to datetime format.
# The errors='coerce' part will automatically turn any problematic dates into 'NaT' (Not a Time), which we can handle as a missing value.
df['Order_Date'] = pd.to_datetime(df['Order_Date'], errors='coerce')

# Now, we'll drop any rows where the date could not be fixed (where Order_Date is NaT)
df.dropna(subset=['Order_Date'], inplace=True)
```

```python
# Fill missing numerical values
df['Units_Sold'].fillna(df['Units_Sold'].median(), inplace=True)
df['Unit_Price'].fillna(df['Unit_Price'].median(), inplace=True)
df['Discount_%'].fillna(0, inplace=True)
df['Sales_Amount'].fillna(0, inplace=True)
```

```python
# Fill missing numerical values
df['Units_Sold'] = df['Units_Sold'].fillna(df['Units_Sold'].median())
df['Unit_Price'] = df['Unit_Price'].fillna(df['Unit_Price'].median())
df['Discount_%'] = df['Discount_%'].fillna(0)
df['Sales_Amount'] = df['Sales_Amount'].fillna(0)
```

```python
# Fill missing categorical values
df['City'] = df['City'].fillna('Unknown')
df['Segment'] = df['Segment'].fillna('Unknown')
```

```python
# Check how many duplicate rows exist
print(f"Number of duplicate rows: {df.duplicated().sum()}")

# Remove duplicate rows by re-assigning the DataFrame
df = df.drop_duplicates()

print(f"Number of rows after dropping duplicates: {len(df)}")
```

```python
# Run the final check
print("\nData summary after cleaning:")
df.info()
```

```python
# Define the file name for our clean data
cleaned_file_name = 'snitch_sales_cleaned.csv'

# Save the cleaned DataFrame to a new CSV file
# index=False prevents pandas from adding an extra 'index' column to your file
df.to_csv(cleaned_file_name, index=False)

print(f"Successfully saved cleaned data to {cleaned_file_name}!")
```
